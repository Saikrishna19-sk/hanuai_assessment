## Maintenance Data Analysis

### Objectives
- Inspect the maintenance dataset.
- Check missing values and duplicate records.
- Categorize issue descriptions.
- Explore text clusters using TF-IDF and KMeans.
- Generate outputs for stakeholder review.

### Limitations
Keyword categories and text clusters are exploratory results. They do not independently establish maintenance root causes.


```python

from pathlib import Path
import pandas as pd
from IPython.display import display

PROJECT_ROOT = Path.cwd()

if not (PROJECT_ROOT / "src").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

INPUT_DIR = PROJECT_ROOT / "data" / "input"
RESULT_FILE = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "task2_maintenance_analysis.csv"
)

print("Project root:", PROJECT_ROOT)
print("Input directory:", INPUT_DIR)
print("Analysis output exists:", RESULT_FILE.exists())

if INPUT_DIR.exists():
    print("\nInput files:")
    for path in sorted(INPUT_DIR.iterdir()):
        if path.is_file():
            print("-", path.name)

if RESULT_FILE.exists():
    results = pd.read_csv(RESULT_FILE)
    print("\nDataset dimensions:", results.shape)
    display(results.head(10))

    if "issue_category" in results.columns:
        print("\nIssue categories:")
        display(results["issue_category"].value_counts().to_frame("count"))

    if "text_cluster" in results.columns:
        print("\nText clusters:")
        display(results["text_cluster"].value_counts().to_frame("count"))
else:
    print("\nRun the Task 2 script to generate the analysis output.")
```
